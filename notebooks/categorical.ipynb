```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/mpg_raw.csv")
df.head()
```

```python
sns.catplot(x='origin', y='horsepower', data=df, kind='bar', hue='cylinders', palette='coolwarm', height=4 );
```

## Insights: Average horsepower by origin and cylinders

- **4 and 6 cylinders appear in all three origins.** The other cylinder counts are not shared: 8 only in USA, 3 only in Japan, 5 only in Europe.
- **More cylinders generally means more horsepower.** In the USA it goes from ~80 hp (4 cyl) to ~100 hp (6 cyl) to ~158 hp (8 cyl).
- **USA 8-cylinder cars have the highest average horsepower** (~158 hp). No other origin has 8-cylinder cars, so the USA's high overall horsepower is partly because of them.
- **4-cylinder cars are similar across origins** (~76-81 hp).
- **6-cylinder cars are weaker in the USA** (~100 hp) than in Japan and Europe (~115 hp).
- **Wide error bars mean low confidence.** Europe's 5-cylinder bar and Japan's 3-cylinder bar have long error bars, which usually means very few cars. Don't draw conclusions from those bars.

## Insights: mpg by origin and cylinders

- **USA cars are the least fuel-efficient.** Most sit at 10-20 mpg, and the bottom of the chart is mostly 8-cylinder cars.
- **Japan and Europe are mostly 4-cylinder cars** at 25-40 mpg.
- **More cylinders means lower mpg** in every origin.

### Best choice
- **Datsun 280-ZX** (Japan, 6 cyl, 32.7 mpg): a 6-cylinder that still gets over 30 mpg.
- **Oldsmobile Cutlass LS** (USA, 8 cyl, over 25 mpg): the only 8-cylinder above 25 mpg.

### Worst choice
- **Mazda RX2 Coupe and Mazda RX3** (Japan, 3 cyl, under 20 mpg): far below the 25-40 mpg typical for Japanese cars.

### Takeaway
Cylinders predict mpg well, so the best and worst picks are the cars that break that pattern.
