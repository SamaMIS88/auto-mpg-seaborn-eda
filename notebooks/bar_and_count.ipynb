```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/mpg_raw.csv")
df.head()
```

```python
g=sns.barplot(x='origin', y='horsepower', data=df,estimator=np.mean,hue='origin', palette='coolwarm')
plt.title('Barplot of Horsepower by Origin');
```

## Insights: horsepower by origin
- **USA cars have the highest average horsepower** (about 119 hp).
- **Japan and Europe are similar** (about 80 hp).

```python
g=sns.countplot(x='origin', data=df, hue='origin', palette='coolwarm')
plt.title('Count of Cars by Origin');
```

## Insights: count by origin
- **USA makes up most of the dataset** (249 of 398 cars), so averages for USA are more reliable than for Japan or Europe.

```python
df['mpg_level']=df['mpg'].apply(lambda x:'low' if x < 15  else 'high' if x > 30 else 'medium' ); 
```

```python
df.head()   
```

```python
sns.countplot(x='model_year', data=df, hue='mpg_level', palette='coolwarm')
plt.title('Count of Cars by Model Year and MPG Level')
plt.Figure(figsize=(10,6));
```

## Insights: mpg level by model year
- **Low-mpg cars (under 15) disappear after 1976.**
- **High-mpg cars (over 30) grow from 1977 and dominate 1980-82** (about 16-19 cars per year).
- **Fuel efficiency improved over time.**

```python
sns.countplot(x='model_year', data=df, hue='cylinders', palette='coolwarm')
plt.title('Count of Cars by Model Year and Cylinders')
plt.figure(figsize=(10,6));
```

## Insights: cylinders by model year
- **8-cylinder cars fade out**, from 18 in 1970 to 0-1 after 1980.
- **4-cylinder cars take over** (28 of 31 cars in 1982).
