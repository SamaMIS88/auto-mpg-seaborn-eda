```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/mpg_raw.csv")
df.head()
```

```python
g=sns.barplot(x='origin', y='horsepower', data=df,estimator=np.mean,hue='origin', palette='coolwarm')
plt.title('Barplot of Horsepower by Origin');
```

```python
g=sns.countplot(x='origin', data=df, hue='origin', palette='coolwarm')
plt.title('Count of Cars by Origin');
```

```python
df['mpg_level']=df['mpg'].apply(lambda x:'low' if x < 15  else 'high' if x > 30 else 'medium' ); 
```

```python
df.head()   
```

```python
sns.countplot(x='model_year', data=df, hue='mpg_level', palette='coolwarm')
plt.title('Count of Cars by Model Year and MPG Level')
plt.Figure(figsize=(10,6));
```

```python
sns.countplot(x='model_year', data=df, hue='cylinders', palette='coolwarm')
plt.title('Count of Cars by Model Year and Cylinders')
plt.Figure(figsize=(10,6)) ;
```
