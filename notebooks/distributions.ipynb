```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/mpg_raw.csv")
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe(include='all')
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df, x='mpg', bins=20, hue='origin', multiple='dodge', alpha=0.5, kde=False)
plt.title('Histogram of mpg');
```

```python
# Japanese and European cars are more economical than American ones.
```

```python
g=sns.jointplot(data=df, x='mpg', y='weight', hue='origin', kind='scatter', alpha=0.5 , height=7)
```

```python
## Heavier cars get lower mpg
```
